# ⚙️ Setup — Run This First


```python
import time, random, collections, heapq
random.seed(42)

def timer(fn, *args, **kwargs):
    """Time a function and return (result, seconds)."""
    t0 = time.perf_counter()
    result = fn(*args, **kwargs)
    return result, time.perf_counter() - t0

def make_orders(n=10_000):
    """Generate n synthetic PizzaRun orders."""
    zones  = [f"Zone_{i}" for i in range(10)]
    pizzas = ["Margherita","Pepperoni","BBQ Chicken","Veggie","Four Cheese",
              "Hawaiian","Mushroom","Truffle","Spicy Nduja","Buffalo"]
    return [{"order_id": f"ORD-{i:05d}", "pizza_type": random.choice(pizzas),
             "order_time": random.randint(0, 23), "zone": random.choice(zones),
             "priority": random.randint(1, 5)} for i in range(n)]

def build_city_graph():
    """10-zone undirected unweighted graph (adjacency list as dict)."""
    edges = [(0,1),(0,2),(1,3),(1,4),(2,4),(2,5),(3,6),(4,6),
             (4,7),(5,7),(5,8),(6,9),(7,9),(8,9)]
    graph = collections.defaultdict(list)
    for u, v in edges:
        graph[u].append(v)
        graph[v].append(u)
    return graph

print("✅ Setup complete — make_orders(), build_city_graph() ready")
```

```python
import collections

# Small 5-zone city graph (adjacency list as a dict)
small_city = {
    0: [1, 2],
    1: [0, 3],
    2: [0, 4],
    3: [1, 4],
    4: [2, 3],
}

def are_connected(graph, start, end):
    """
    Return True if there is any path from start to end.
    Uses BFS with a queue (collections.deque).
    Time: O(V+E)  |  Space: O(V)
    """
    if start == end:
        return True

    visited = set()
    visited.add(start)
    queue = collections.deque([start])

    while queue:
        node = queue.popleft()   # take from the front (FIFO)
        for neighbour in graph[node]:
            if neighbour == end:
                return True
            if neighbour not in visited:
                # TODO: mark neighbour as visited and add to queue
                pass

    return False  # never reached end

def bfs_hops(graph, start, end):
    """
    Return the minimum number of hops from start to end.
    Return -1 if not reachable.
    Time: O(V+E)  |  Space: O(V)
    """
    if start == end:
        return 0

    visited = {start}
    queue   = collections.deque([(start, 0)])  # (node, hops so far)

    while queue:
        node, hops = queue.popleft()
        for neighbour in graph[node]:
            if neighbour == end:
                return hops + 1
            if neighbour not in visited:
                visited.add(neighbour)
                # TODO: add (neighbour, hops + 1) to the queue
                pass

    return -1  # not reachable


print("0 connected to 4?", are_connected(small_city, 0, 4))  # True
print("Hops from 0 to 4:", bfs_hops(small_city, 0, 4))       # 2

# Now test with the full 10-zone city
city = build_city_graph()
print(f"\n10-zone city: 0 → 9 reachable? {are_connected(city, 0, 9)}")
print(f"Min hops 0 → 9: {bfs_hops(city, 0, 9)}")
```
